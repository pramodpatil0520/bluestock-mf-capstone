```python
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

from pathlib import Path

import plotly.io as pio
```

```python
DATA_PATH = Path("../data/processed/new_processed")

fund_master = pd.read_csv(DATA_PATH/"clean_fund_master.csv")
nav = pd.read_csv(DATA_PATH/"clean_nav_history.csv")
aum = pd.read_csv(DATA_PATH/"clean_aum.csv")
sip = pd.read_csv(DATA_PATH/"clean_sip.csv")
category = pd.read_csv(DATA_PATH/"clean_category.csv")
folio = pd.read_csv(DATA_PATH/"clean_folio.csv")
transactions = pd.read_csv(DATA_PATH/"clean_transactions.csv")
holdings = pd.read_csv(DATA_PATH/"clean_holdings.csv")
```

```python
datasets = {
    "Fund Master": fund_master,
    "NAV": nav,
    "AUM": aum,
    "SIP": sip,
    "Category": category,
    "Folio": folio,
    "Transactions": transactions,
    "Holdings": holdings
}

for name, df in datasets.items():
    print("="*60)
    print(name)
    print(df.shape)
    display(df.head())
```

```python
nav["date"] = pd.to_datetime(nav["date"])

top_funds = nav["amfi_code"].unique()[:10]

nav_sample = nav[
    nav["amfi_code"].isin(top_funds)
]

fig = px.line(
    nav_sample,
    x="date",
    y="nav",
    color="amfi_code",
    title="NAV Trend Analysis (Top 10 Funds)"
)

fig.show()
```

```python
top_aum = (
    aum.groupby("fund_house")["aum_crore"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

fig = px.bar(
    top_aum,
    x="aum_crore",
    y="fund_house",
    orientation="h",
    title="Top 10 Fund Houses by AUM",
    text_auto=True
)

fig.show()
```

```python
sip["month"] = pd.to_datetime(sip["month"])

fig = px.line(
    sip,
    x="month",
    y="sip_inflow_crore",
    markers=True,
    title="Monthly SIP Inflow Trend"
)

fig.show()
```

```python
pivot = category.pivot_table(
    values="net_inflow_crore",
    index="category",
    columns="month",
    aggfunc="sum"
)

fig = px.imshow(
    pivot,
    aspect="auto",
    title="Category-wise Inflow Heatmap"
)

fig.show()
```

```python
age_data = (
    transactions["age_group"]
    .value_counts()
    .reset_index()
)

fig = px.pie(
    age_data,
    names="age_group",
    values="count",
    hole=0.5,
    title="Investor Age Distribution"
)

fig.show()
```

```python
fig = px.box(
    transactions,
    x="age_group",
    y="amount_inr",
    color="age_group",
    title="Investment Amount by Age Group"
)

fig.show()
```

```python
state_data = (
    transactions.groupby("state")
    ["amount_inr"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    state_data,
    x="amount_inr",
    y="state",
    orientation="h",
    title="State-wise Investment Distribution"
)

fig.show()
```

```python
tier_data = (
    transactions["city_tier"]
    .value_counts()
    .reset_index()
)

fig = px.pie(
    tier_data,
    names="city_tier",
    values="count",
    hole=0.5,
    title="T30 vs B30 Investor Distribution"
)

fig.show()
```

```python
fig = px.line(
    folio,
    x=folio.columns[0],
    y=folio.columns[1],
    markers=True,
    title="Industry Folio Growth"
)

fig.show()
```

```python
pivot_nav = nav.pivot_table(
    index="date",
    columns="amfi_code",
    values="nav"
)

returns = pivot_nav.pct_change()

corr = returns.corr()

fig = px.imshow(
    corr,
    title="Mutual Fund Correlation Matrix"
)

fig.show()
```

```python
sector_data = (
    holdings.groupby("sector")
    ["weight_pct"]
    .sum()
    .reset_index()
)

fig = px.pie(
    sector_data,
    names="sector",
    values="weight_pct",
    hole=0.5,
    title="Sector Allocation"
)

fig.show()
```

```python
top_states = (
    transactions.groupby("state")
    ["amount_inr"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_states)
```

```python
top_funds = (
    transactions.groupby("amfi_code")
    ["amount_inr"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_funds)
```
